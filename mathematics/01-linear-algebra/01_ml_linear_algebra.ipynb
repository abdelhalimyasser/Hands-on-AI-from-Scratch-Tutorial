## Vector Addition

Two vectors can be added if they have the same dimension.

If:

u = [u1 u2 ... un]
v = [v1 v2 ... vn]

then:

u + v = [u1+v1, u2+v2, ..., un+vn]

So vector addition is performed element by element.

In our dataset, since both students are represented by 5-dimensional vectors,
we can add them directly.

For visualization, we will also plot the first two components in 2D.

```octave
%% Vector Addition

disp("Student1 first vector\n")
student1

disp("\nStudent2 second vector\n")
student2

disp("\n\nSummation to both of two vectors\n")
sum = student1 + student2
```

## Vector Subtraction

Vector subtraction is also performed element by element.

If:

u = [u1 u2 ... un]
v = [v1 v2 ... vn]

then:

u - v = [u1-v1, u2-v2, ..., un-vn]

Geometrically, subtraction can be understood as:

u - v = u + (-v)

So subtracting a vector is the same as adding its opposite.

```octave
%% Vector Substraction

disp("Student1 first vector\n")
student1

disp("\nStudent2 second vector\n")
student2

disp("\n\nSubstaction to both of two vectors\n")
diff = student1 - student2
```

## Vector Scaling

Vector scaling means multiplying a vector by a scalar number.

If:

v = [v1 v2 ... vn]

and c is a scalar, then:

cv = [cv1 cv2 ... cvn]

Scaling changes the magnitude of the vector.

- If c > 1, the vector becomes longer.
- If 0 < c < 1, the vector becomes shorter.
- If c < 0, the direction is reversed.

```octave
%% Vector Scaling

disp("\nVector Scaled Up\n")
scaledUp   = 2 * student1

disp("\nVector Scaled Down\n")
scaledDown = 0.5 * student1

disp("\nVector Scaled to Negative (reversed)\n")
scaledNeg  = -1 * student1
```

```octave
%% Plot: Vector Scaling

figure;
hold on;
grid on;
axis equal;

scaledUp2D   = 2 * v1;
scaledDown2D = 0.5 * v1;
scaledNeg2D  = -1 * v1;

quiver(0,0,v1(1),v1(2),0,'LineWidth',2);
quiver(0,0,scaledUp2D(1),scaledUp2D(2),0,'LineWidth',2);
quiver(0,0,scaledDown2D(1),scaledDown2D(2),0,'LineWidth',2);
quiver(0,0,scaledNeg2D(1),scaledNeg2D(2),0,'LineWidth',2);

text(v1(1), v1(2), '  v1');
text(scaledUp2D(1), scaledUp2D(2), '  2v1');
text(scaledDown2D(1), scaledDown2D(2), '  0.5v1');
text(scaledNeg2D(1), scaledNeg2D(2), '  -v1');

xlabel('Study Hours');
ylabel('Attendance');
title('Vector Scaling');
legend('v1','2v1','0.5v1','-v1','Location','northeast');
hold off;
```

### Interpretation

Scaling changes the length of the vector.

- Multiplying by 2 doubles the vector length.
- Multiplying by 0.5 halves the vector length.
- Multiplying by -1 keeps the same length but reverses the direction.

Therefore, scalar multiplication controls both:

- magnitude
- direction

# Vector Norms

A **norm** is a mathematical function used to measure the **magnitude (length)** of a vector.

If we have a vector:

$$
\mathbf{v} =
\begin{bmatrix}
v_1 \\
v_2 \\
\vdots \\
v_n
\end{bmatrix}
$$

then a norm takes this vector as input and returns **one non-negative scalar value**:

$$
\mathbf{v}
\quad \longrightarrow \quad
\|\mathbf{v}\|
$$

So:

$$
\text{Vector} \rightarrow \text{Norm} \rightarrow \text{Scalar}
$$

For example:

$$
\mathbf{v} =
\begin{bmatrix}
3 \\
4
\end{bmatrix}
$$

A norm can be used to describe how "large" this vector is.

---

## Why do we need norms?

Vectors may contain many components.

For example, in our student dataset:

$$
\mathbf{s}_1 =
\begin{bmatrix}
4.8 & 53 & 78 & 70 & 80
\end{bmatrix}
$$

This vector contains five values.

A norm allows us to summarize the magnitude of this vector using
a single number:

$$
\|\mathbf{s}_1\|
$$

Different norms measure the magnitude in different ways.

Therefore, the same vector can have different norm values depending
on the norm function used.

---

# Main Uses of Norms

According to the lecture, norms are used for several important purposes.

## 1. Measuring Distance

Norms can be used to measure the distance between two vectors.

If we have two vectors:

$$
\mathbf{u}
\quad\text{and}\quad
\mathbf{v}
$$

first calculate the difference:

$$
\mathbf{d} = \mathbf{u} - \mathbf{v}
$$

then calculate its norm:

$$
\text{Distance} = \|\mathbf{u} - \mathbf{v}\|
$$

So the general idea is:

$$
\mathbf{u}
\longrightarrow
\mathbf{u}-\mathbf{v}
\longrightarrow
\|\mathbf{u}-\mathbf{v}\|
$$

Different norms produce different definitions of distance.

---

## 2. Normalization

Different variables may have very different scales.

For example:

$$
\text{Study Hours} \approx 1 \text{ to } 7
$$

while:

$$
\text{Attendance} \approx 50 \text{ to } 100
$$

A large-scale variable may dominate calculations.

Norms can be used to rescale vectors so that their magnitudes become comparable.

A common idea is:

$$
\hat{\mathbf{v}}
=
\frac{\mathbf{v}}
{\|\mathbf{v}\|}
$$

The resulting vector:

$$
\hat{\mathbf{v}}
$$

has magnitude equal to:

$$
\|\hat{\mathbf{v}}\| = 1
$$

This is called a **unit vector**.

---

## 3. Regularization

Norms can also be used as penalty terms.

The idea is to prevent values from becoming unnecessarily large.

For example:

$$
\text{Objective}
=
\text{Error}
+
\lambda \|\mathbf{w}\|
$$

where:

- $\mathbf{w}$ represents some parameters.
- $\lambda$ controls the strength of the penalty.
- $\|\mathbf{w}\|$ measures the size of the parameters.

Different norms create different types of penalties.

---

## 4. Similarity Measurement

Norms can help measure how similar two vectors are.

For example, the angle between two vectors can be calculated using:

$$
\cos(\theta)
=
\frac{\mathbf{u}\cdot\mathbf{v}}
{\|\mathbf{u}\|\|\mathbf{v}\|}
$$

The norms in the denominator are used to remove the effect of vector magnitude.

This allows us to focus more on direction.

---

## 5. Optimization

Norms are also used to measure errors.

For example, if:

$$
\mathbf{e}
=
\mathbf{y}
-
\hat{\mathbf{y}}
$$

then the norm of the error vector can describe how large the total error is:

$$
\|\mathbf{e}\|
$$

Different norms give different error measurements.

For example:

$$
L_1 \rightarrow \text{Absolute Error}
$$

and:

$$
L_2 \rightarrow \text{Squared Error}
$$

---

## 6. Data Representation and Feature Extraction

Norms are also useful when working with high-dimensional data.

They can help measure:

- vector magnitude
- distances
- errors
- similarity
- important directions in data

These ideas are used later in techniques such as dimensionality reduction.

## L1 Norm (1-Norm)

The **L1 norm** measures the magnitude of a vector by adding the
absolute values of all its components.

For a vector:

$$
\mathbf{v}
=
\begin{bmatrix}
v_1 \\
v_2 \\
\vdots \\
v_n
\end{bmatrix}
$$

the L1 norm is defined as:

$$
\|\mathbf{v}\|_1
=
|v_1| + |v_2| + \cdots + |v_n|
$$

or more compactly:

$$
\|\mathbf{v}\|_1
=
\sum_{i=1}^{n} |v_i|
$$

The absolute value is important because negative values should not
cancel positive values.

The result is always a non-negative scalar.

```octave
%% L1 Norm - Manual Calculation

clear sum

disp("Student 1 Vector:")
student1

absolute_values = abs(student1)

disp("\nManual L1 Norm:")
l1_manual = sum(absolute_values)

%% Verify using norm()

disp("\nMATLAB L1 Norm:")
l1_matlab = norm(student1, 1)
```

### L1 Norm Application: Measuring Distance

The L1 norm can also be used to measure the distance between two vectors.

For two vectors:

$$
\mathbf{u}
\quad\text{and}\quad
\mathbf{v}
$$

we first calculate their difference:

$$
\mathbf{d}
=
\mathbf{u} - \mathbf{v}
$$

Then the L1 distance is:

$$
\|\mathbf{u} - \mathbf{v}\|_1
$$

This is also called **Manhattan Distance**.

The idea is to add the absolute differences between corresponding components.

```octave
%% Measuring Distance using L1 Norm


% Step 1: Subtract the two vectors
difference = student1 - student2;

disp("Difference:")
difference

% Step 2: Take absolute values
absolute_difference = abs(difference);

disp("\n\nAbsolute Difference:")
absolute_difference

% Step 3: Sum the absolute differences
l1_distance_manual = sum(absolute_difference);

disp("\nManual L1 Distance:")
l1_distance_manual

%% Verify L1 Distance using norm()

l1_distance_matlab = norm(student1 - student2, 1);

disp("\nL1 Distance using norm():")
l1_distance_matlab
```

#### Visualizing L1 Distance

To visualize Manhattan Distance, we use two 2D points.
The L1 distance is calculated by moving horizontally and vertically.

```octave
%% Plot L1 Manhattan Distance Between Two Students

% Take first two dimensions from the same students
% X = Study Hours
% Y = Attendance

p1 = student1(1:2);
p2 = student2(1:2);

% L1 distance using these two dimensions
l1_2d = sum(abs(p1 - p2));

figure;
hold on;
grid on;

% Student 1 point
plot(p1(1), p1(2), 'o', ...
    'MarkerSize', 10, ...
    'LineWidth', 2);

% Student 2 point
plot(p2(1), p2(2), 'o', ...
    'MarkerSize', 10, ...
    'LineWidth', 2);

% Manhattan Path
% First move horizontally
plot([p1(1) p2(1)], ...
     [p1(2) p1(2)], ...
     '--', ...
     'LineWidth', 2);

% Then move vertically
plot([p2(1) p2(1)], ...
     [p1(2) p2(2)], ...
     '--', ...
     'LineWidth', 2);

% Labels
text(p1(1), p1(2), ...
    '  Student 1');

text(p2(1), p2(2), ...
    '  Student 2');

% Show horizontal distance
text((p1(1)+p2(1))/2, ...
     p1(2)-2, ...
     ['|', num2str(p2(1)), ...
      ' - ', num2str(p1(1)), ...
      '| = ', num2str(abs(p2(1)-p1(1)))]);

% Show vertical distance
text(p2(1)+0.05, ...
     (p1(2)+p2(2))/2, ...
     ['|', num2str(p2(2)), ...
      ' - ', num2str(p1(2)), ...
      '| = ', num2str(abs(p2(2)-p1(2)))]);

xlabel('Study Hours');
ylabel('Attendance');

title(['L1 Manhattan Distance = ', num2str(l1_2d)]);

legend('Student 1', ...
       'Student 2', ...
       'Horizontal Distance', ...
       'Vertical Distance', ...
       'location','northeast');

hold off;
```

### L1 Norm Application:  Mean Absolute Error

The lecture also connects the L1 idea with
**Mean Absolute Error (MAE)**.

Suppose we have two sets of values:

$$
\mathbf{y}
=
\begin{bmatrix}
y_1 & y_2 & \cdots & y_n
\end{bmatrix}
$$

and:

$$
\hat{\mathbf{y}}
=
\begin{bmatrix}
\hat{y}_1 & \hat{y}_2 & \cdots & \hat{y}_n
\end{bmatrix}
$$

The error vector is:

$$
\mathbf{e}
=
\mathbf{y}
-
\hat{\mathbf{y}}
$$

MAE calculates the average absolute error:

$$
MAE
=
\frac{1}{n}
\sum_{i=1}^{n}
|y_i - \hat{y}_i|
$$

So MAE uses the same absolute-value idea as the L1 norm.

```octave
%% Mean Absolute Error Example

actual = [80 90 70 60];
estimated = [78 94 68 65];

errors = actual - estimated;

absolute_errors = abs(errors);

MAE = mean(absolute_errors);

disp("Errors:")
errors

disp("Absolute Errors:")
absolute_errors

disp("MAE:")
MAE
```

```octave
%% MAE Step by Step

actual = [80 90 70 60];
estimated = [78 94 68 65];

errors = actual - estimated

absolute_errors = abs(errors)

sum_absolute_errors = sum(absolute_errors)

number_of_values = length(actual)

MAE_manual = sum_absolute_errors / number_of_values
```

### MAE Steps

The calculation is:

$$
\mathbf{e}
=
\begin{bmatrix}
80-78 &
90-94 &
70-68 &
60-65
\end{bmatrix}
$$

So:

$$
\mathbf{e}
=
\begin{bmatrix}
2 & -4 & 2 & -5
\end{bmatrix}
$$

Take absolute values:

$$
|\mathbf{e}|
=
\begin{bmatrix}
2 & 4 & 2 & 5
\end{bmatrix}
$$

Then calculate the mean:

$$
MAE
=
\frac{2+4+2+5}{4}
$$

$$
=
\frac{13}{4}
$$

$$
=
3.25
$$

## L2 Norm (2-Norm)

The **L2 norm** measures the magnitude of a vector by taking the
square root of the sum of the squared values of all its components.

For a vector:

$$
\mathbf{v}
=
\begin{bmatrix}
v_1 \\
v_2 \\
\vdots \\
v_n
\end{bmatrix}
$$

the L2 norm is defined as:

$$
\|\mathbf{v}\|_2
=
\sqrt{
v_1^2 + v_2^2 + \cdots + v_n^2
}
$$

or more compactly:

$$
\|\mathbf{v}\|_2
=
\sqrt{
\sum_{i=1}^{n} v_i^2
}
$$

Squaring the values makes all components non-negative.

The result is always a non-negative scalar.

The L2 norm represents the ordinary geometric length
of a vector.

```octave
%% L2 Norm - Manual Calculation

clear sum

disp("Student 1 Vector:")
student1

squared_values = student1.^2

disp("\nManual L2 Norm:")
l2_manual = sqrt(sum(squared_values))

%% Verify using norm()

disp("\nMATLAB L2 Norm:")
l2_matlab = norm(student1, 2)
```

### L2 Norm Application: Measuring Distance

The L2 norm can also be used to measure the distance between two vectors.

For two vectors:

$$
\mathbf{u}
\quad\text{and}\quad
\mathbf{v}
$$

we first calculate their difference:

$$
\mathbf{d}
=
\mathbf{u} - \mathbf{v}
$$

Then the L2 distance is:

$$
\|\mathbf{u} - \mathbf{v}\|_2
$$

This is also called **Euclidean Distance**.

The idea is to square the differences between corresponding components,
add them together, and then take the square root.

```octave
%% Measuring Distance using L2 Norm

% Step 1: Subtract the two vectors
difference = student1 - student2;

disp("Difference:")
difference

% Step 2: Square the differences
squared_difference = difference.^2;

disp("\n\nSquared Difference:")
squared_difference

% Step 3: Sum the squared differences
sum_squared_difference = sum(squared_difference);

disp("\nSum of Squared Differences:")
sum_squared_difference

% Step 4: Take the square root
l2_distance_manual = sqrt(sum_squared_difference);

disp("\nManual L2 Distance:")
l2_distance_manual

%% Verify L2 Distance using norm()

l2_distance_matlab = norm(student1 - student2, 2);

disp("\nL2 Distance using norm():")
l2_distance_matlab
```

### L2 Norm Application: Root Mean Square Error

The lecture also connects the L2 idea with
**Root Mean Square Error (RMSE)**.

Suppose we have two sets of values:

$$
\mathbf{y}
=
\begin{bmatrix}
y_1 & y_2 & \cdots & y_n
\end{bmatrix}
$$

and:

$$
\hat{\mathbf{y}}
=
\begin{bmatrix}
\hat{y}_1 & \hat{y}_2 & \cdots & \hat{y}_n
\end{bmatrix}
$$

The error vector is:

$$
\mathbf{e}
=
\mathbf{y}
-
\hat{\mathbf{y}}
$$

RMSE squares each error, calculates the mean,
and then takes the square root:

$$
RMSE
=
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

So RMSE uses the same square-and-square-root idea as the L2 norm.

```octave
%% Root Mean Square Error Example

actual = [80 90 70 60];
estimated = [78 94 68 65];

errors = actual - estimated;

squared_errors = errors.^2;

RMSE = sqrt(mean(squared_errors));

disp("Errors:")
errors

disp("Squared Errors:")
squared_errors

disp("RMSE:")
RMSE
```

```octave
%% RMSE Step by Step

actual = [80 90 70 60];
estimated = [78 94 68 65];

errors = actual - estimated

squared_errors = errors.^2

sum_squared_errors = sum(squared_errors)

number_of_values = length(actual)

mean_squared_error = sum_squared_errors / number_of_values

RMSE_manual = sqrt(mean_squared_error)
```

### RMSE Steps

The calculation is:

$$
\mathbf{e}
=
\begin{bmatrix}
80-78 &
90-94 &
70-68 &
60-65
\end{bmatrix}
$$

So:

$$
\mathbf{e}
=
\begin{bmatrix}
2 & -4 & 2 & -5
\end{bmatrix}
$$

Square the errors:

$$
\mathbf{e}^2
=
\begin{bmatrix}
4 & 16 & 4 & 25
\end{bmatrix}
$$

Then calculate the mean:

$$
MSE
=
\frac{4+16+4+25}{4}
$$

$$
=
\frac{49}{4}
$$

$$
=
12.25
$$

Finally, take the square root:

$$
RMSE
=
\sqrt{12.25}
$$

$$
=
3.5
$$

## Max Norm (L-Infinity Norm)

The **Max Norm** measures the magnitude of a vector by taking the
largest absolute value among all its components.

For a vector:

$$
\mathbf{v}
=
\begin{bmatrix}
v_1 \\
v_2 \\
\vdots \\
v_n
\end{bmatrix}
$$

the Max Norm is defined as:

$$
\|\mathbf{v}\|_\infty
=
\max
\left(
|v_1|,
|v_2|,
\ldots,
|v_n|
\right)
$$

The absolute value is important because we are interested in
magnitude, not direction.

The result is the largest magnitude among all components.

The result is always a non-negative scalar.

```octave
%% Max Norm - Manual Calculation

clear max

disp("Student 1 Vector:")
student1

absolute_values = abs(student1)

disp("\nManual Max Norm:")
max_manual = max(absolute_values)

%% Verify using norm()

disp("\nMATLAB Max Norm:")
max_matlab = norm(student1, inf)
```

$$

# Dot product

## Dot Product

The **dot product** is a way to multiply two vectors of the same length.

For two vectors:

$$
\mathbf{u}
=
\begin{bmatrix}
u_1 \\
u_2 \\
\vdots \\
u_n
\end{bmatrix}
$$

and:

$$
\mathbf{v}
=
\begin{bmatrix}
v_1 \\
v_2 \\
\vdots \\
v_n
\end{bmatrix}
$$

the dot product is calculated by multiplying the corresponding
elements and then adding the results.

$$
\mathbf{u}\cdot\mathbf{v}
=
u_1v_1 + u_2v_2 + \cdots + u_nv_n
$$

or:

$$
\boxed{
\mathbf{u}\cdot\mathbf{v}
=
\sum_{i=1}^{n}
u_i v_i
}
$$

The two vectors must have the same number of elements.

The result of the dot product is one scalar value.

```octave
%% Dot Product - Manual Calculation

clear sum

disp("Student 1 Vector:")
student1

disp("\nStudent 2 Vector:")
student2

% Step 1: Multiply corresponding elements
element_products = student1 .* student2

% Step 2: Sum all products
disp("\nManual Dot Product:")
dot_manual = sum(element_products)

%% Verify using dot()

disp("\nMATLAB Dot Product:")
dot_matlab = dot(student1, student2)
```

### Dot Product Steps

For:

$$
\mathbf{s}_1
=
\begin{bmatrix}
4.8 & 53 & 78 & 70 & 80
\end{bmatrix}
$$

and:

$$
\mathbf{s}_2
=
\begin{bmatrix}
5.4 & 86 & 94 & 96 & 100
\end{bmatrix}
$$

the dot product is:

$$
\mathbf{s}_1\cdot\mathbf{s}_2
=
(4.8)(5.4)
+
(53)(86)
+
(78)(94)
+
(70)(96)
+
(80)(100)
$$

The corresponding elements are multiplied first.

Then all products are added together.

The final result is one scalar value.

## Geometric Meaning of the Dot Product

The dot product can also be written using the angle between two vectors:

$$
\mathbf{a}\cdot\mathbf{b}
=
\|\mathbf{a}\|
\|\mathbf{b}\|
\cos(\theta)
$$

where:

- $\|\mathbf{a}\|$ is the magnitude of vector $\mathbf{a}$
- $\|\mathbf{b}\|$ is the magnitude of vector $\mathbf{b}$
- $\theta$ is the angle between the two vectors

This means that the dot product depends on:

1. The magnitude of both vectors.
2. The angle between them.

The angle therefore tells us something about the directions
of the two vectors.

## Projection Product

To understand projection, we first convert vector $\mathbf{b}$
into a unit vector.

A **unit vector** has magnitude equal to:

$$
1
$$

The unit vector in the direction of $\mathbf{b}$ is:

$$
\hat{\mathbf{b}}
=
\frac{\mathbf{b}}
{\|\mathbf{b}\|_2}
$$

This keeps the same direction of $\mathbf{b}$,
but changes its length to 1.

```octave
%% Unit Vector of b

b = student2(1:2);

disp("Vector b:")
b

% Step 1: Calculate magnitude
b_magnitude = norm(b, 2)

% Step 2: Create unit vector
unit_b = b / b_magnitude

% Verify magnitude
disp("\nMagnitude of Unit Vector:")
unit_b_magnitude = norm(unit_b, 2)
```

### Projection of a onto b

Suppose we have two vectors:

$$
\mathbf{a}
\quad\text{and}\quad
\mathbf{b}
$$

First, convert $\mathbf{b}$ into its unit vector:

$$
\hat{\mathbf{b}}
=
\frac{\mathbf{b}}
{\|\mathbf{b}\|_2}
$$

Then calculate:

$$
\boxed{
\mathbf{a}\cdot\hat{\mathbf{b}}
}
$$

This gives the projection of $\mathbf{a}$
in the direction of $\mathbf{b}$.

The result is a scalar value.

```octave
%% Projection Product

a = student1(1:2);
b = student2(1:2);

disp("Vector a:")
a

disp("\nVector b:")
b

% Step 1: Unit vector of b
unit_b = b / norm(b, 2)

disp("\nUnit Vector of b:")
unit_b

% Step 2: Dot product
projection = dot(a, unit_b)

disp("\nProjection of a in direction of b:")
projection
```

```octave
%% Plot Projection Product

a = student1(1:2);
b = student2(1:2);

unit_b = b / norm(b, 2);

scalar_projection = dot(a, unit_b);

projection_vector = scalar_projection * unit_b;

figure;
hold on;
grid on;
axis equal;

% Vector a
quiver(0,0,a(1),a(2),0,'LineWidth',2);

% Vector b
quiver(0,0,b(1),b(2),0,'LineWidth',2);

% Projection
quiver(0,0,projection_vector(1),projection_vector(2),0,'LineWidth',2);

% Perpendicular line
plot([a(1) projection_vector(1)], ...
     [a(2) projection_vector(2)], ...
     '--','LineWidth',1.5);

text(a(1),a(2),'  a');
text(b(1),b(2),'  b');
text(projection_vector(1), ...
     projection_vector(2), ...
     '  Projection');

xlabel('Study Hours');
ylabel('Attendance');

title('Projection of a onto b');

legend('Vector a', ...
       'Vector b', ...
       'Projection', ...
       'Perpendicular', ...
       'location','northeast');

hold off;
```

## Cosine Similarity

**Cosine Similarity** measures the similarity between two vectors
using the angle between them.

It is calculated as:

$$
\boxed{
\cos(\theta)
=
\frac{
\mathbf{a}\cdot\mathbf{b}
}{
\|\mathbf{a}\|_2
\|\mathbf{b}\|_2
}
}
$$

The numerator is the dot product:

$$
\mathbf{a}\cdot\mathbf{b}
$$

The denominator contains the magnitudes of the two vectors:

$$
\|\mathbf{a}\|_2
\|\mathbf{b}\|_2
$$

Cosine similarity focuses on the **direction**
of the vectors rather than only their magnitudes.

```octave
%% Cosine Similarity

a = student1;
b = student2;

% Step 1: Dot product
dot_ab = dot(a, b)

% Step 2: Magnitude of a
magnitude_a = norm(a, 2)

% Step 3: Magnitude of b
magnitude_b = norm(b, 2)

% Step 4: Cosine similarity
cosine_similarity = ...
    dot_ab / (magnitude_a * magnitude_b)

disp("\nCosine Similarity:")
cosine_similarity
```

```octave
%% Cosine Similarity - Manual Calculation

clear sum

a = student1;
b = student2;

% Dot product manually
element_products = a .* b;
dot_manual = sum(element_products)

% Magnitude of a
a_squared = a.^2;
magnitude_a_manual = sqrt(sum(a_squared))

% Magnitude of b
b_squared = b.^2;
magnitude_b_manual = sqrt(sum(b_squared))

% Cosine similarity
cosine_manual = ...
    dot_manual / ...
    (magnitude_a_manual * magnitude_b_manual)

disp("\nManual Cosine Similarity:")
cosine_manual
```

### Interpreting Cosine Similarity

Cosine similarity depends on the angle between two vectors.

If:

$$
\theta = 0^\circ
$$

then:

$$
\cos(\theta)=1
$$

The vectors point in the same direction.

If:

$$
\theta = 90^\circ
$$

then:

$$
\cos(\theta)=0
$$

The vectors are perpendicular.

If:

$$
\theta = 180^\circ
$$

then:

$$
\cos(\theta)=-1
$$

The vectors point in opposite directions.

```octave
%% Visualize Cosine Similarity

a = student1(1:2);
b = student2(1:2);

cosine_similarity_2d = ...
    dot(a,b) / ...
    (norm(a,2) * norm(b,2));

angle_radians = acos(cosine_similarity_2d);

angle_degrees = ...
    angle_radians * 180 / pi;

figure;
hold on;
grid on;
axis equal;

quiver(0,0,a(1),a(2),0,'LineWidth',2);
quiver(0,0,b(1),b(2),0,'LineWidth',2);

text(a(1),a(2),'  Student 1');
text(b(1),b(2),'  Student 2');

xlabel('Study Hours');
ylabel('Attendance');

title([
    'Cosine Similarity = ', ...
    num2str(cosine_similarity_2d), ...
    ' | Angle = ', ...
    num2str(angle_degrees), ...
    ' degrees'
]);

legend('Student 1', ...
       'Student 2', ...
       'location','northeast');

hold off;
```
